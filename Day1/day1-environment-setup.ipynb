## Linux Development Environment Setup

This cell installs core tools, Python (via **pyenv**), **UV**, **Docker**, and **VS Code**.

```python
%%bash
# Install essential tools
sudo apt update && sudo apt upgrade -y
sudo apt install -y git curl wget build-essential zsh

# Install Python using pyenv
curl https://pyenv.run | bash
echo 'export PYENV_ROOT="$HOME/.pyenv"' >> ~/.zshrc
echo 'command -v pyenv >/dev/null || export PATH="$PYENV_ROOT/bin:$PATH"' >> ~/.zshrc
echo 'eval "$(pyenv init -)"' >> ~/.zshrc
source ~/.zshrc

# Install Python
pyenv install 3.11.5
pyenv global 3.11.5

# Install UV (modern Python package manager)
curl -LsSf https://astral.sh/uv/install.sh | sh
echo 'export PATH="$HOME/.cargo/bin:$PATH"' >> ~/.zshrc
source ~/.zshrc

# Install Docker
curl -fsSL https://get.docker.com -o get-docker.sh
sudo sh get-docker.sh
sudo usermod -aG docker "$USER"

# Install VS Code (via snap)
sudo snap install --classic code
```

## Project Structure Setup

This cell creates the **inventory-system** project and scaffolds the directory structure using **UV** and standard folders for source code, tests, deployments, and docs.

```python
%%bash
# Create project directory
mkdir -p ~/projects/inventory-system
cd ~/projects/inventory-system

# Initialize with UV
uv init --package inventory-system
cd inventory-system

# Project structure
mkdir -p src/inventory_system/{models,services,repositories,api,config}
mkdir -p tests/{unit,integration}
mkdir -p scripts
mkdir -p deployments/{docker,kubernetes}
mkdir -p docs
```
